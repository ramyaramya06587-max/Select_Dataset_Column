```python
import pandas as pd

# Load the Iris dataset
df = pd.read_csv("iris.csv")

# Display the first five records
print("First five records:")
print(df.head())
```

```python
# Select specific columns using loc
print("Selected columns using loc:")
print(df.loc[:, ["sepal_length", "species"]])
```

```python
# Select specific columns using iloc
print("Selected columns using iloc:")
print(df.iloc[:, [0, 4]])
```

```python
# Select specific rows using loc
print("First five rows using loc:")
print(df.loc[0:4, :])
```

```python
# Select specific rows using iloc
print("First five rows using iloc:")
print(df.iloc[0:5, :])
```

```python
# Select specific rows and columns together
print("Selected rows and columns:")
print(df.loc[0:4, ["sepal_length", "species"]])
```
